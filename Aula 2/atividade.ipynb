## 📦 Preparação

Execute o código fornecido para carregar sua base personalizada.

```python
import shutil
import urllib.request
from pathlib import Path

import pandas as pd

# Cada pessoa trabalha com uma amostra própria de 800 linhas do Titanic.
# O sorteio parte do seu nome, então a sua amostra é sempre a mesma — inclusive
# quando você abrir a Aula 3.
nome = input('Digite seu nome: ').strip().lower().replace(' ', '')

# data_raw/ guarda os originais (referência) e data/ é a cópia de trabalho.
# Bagunçou a base? Apague o arquivo em data/ e rode esta célula de novo.
Path('data_raw').mkdir(exist_ok=True)
Path('data').mkdir(exist_ok=True)

url = 'https://raw.githubusercontent.com/mwaskom/seaborn-data/master/titanic.csv'
if not Path('data_raw/titanic_completo.csv').exists():
    print('Baixando a base do Titanic...')
    urllib.request.urlretrieve(url, 'data_raw/titanic_completo.csv')

original = Path(f'data_raw/titanic_{nome}.csv')
base = Path(f'data/titanic_{nome}.csv')

if not original.exists():
    semente = sum(ord(letra) for letra in nome)
    amostra = pd.read_csv('data_raw/titanic_completo.csv')
    amostra = amostra.sample(n=800, random_state=semente).reset_index(drop=True)
    amostra.to_csv(original, index=False)
if not base.exists():
    shutil.copy(original, base)

df = pd.read_csv(base)

print(f"Base de {nome} carregada com sucesso!")
print(f'tamanho: {len(df)} linhas')
```
